```python
#percentage of people who know python
numofrespondent = df['Country'].value_counts()
numofrespondent
```

```python
numofknowspython = country_group['LanguageHaveWorkedWith'].apply(lambda x : x.str.contains('Python').sum())
```

```python
numofknowspython
```

```python
knowpy_df = pd.concat([numofrespondent,numofknowspython],axis = 'columns')
```

```python
knowpy_df
```

```python
knowpy_df.rename(columns = {'count':'NumRespondent','LanguageHaveWorkedWith':'KnowPython'}, inplace= True)
```

```python
knowpy_df['pctknowpy'] = (knowpy_df['KnowPython']/knowpy_df['NumRespondent'])*100
```

```python
knowpy_df.columns
```

```python
knowpy_df.sort_values(by='pctknowpy',ascending = False,inplace= True)
```
