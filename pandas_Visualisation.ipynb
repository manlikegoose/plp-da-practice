## Visualisation in Pandas

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

```

```python
df = pd.read_csv('/Ice Cream Ratings.csv',index_col='Date')
```

```python
df
```

```python
df.plot(kind = 'line',title='Ice cream Ratings',xlabel='Time Duration',ylabel='scores')
```

```python
df.plot(kind = 'bar',title='Ice cream Ratings',xlabel='Time Duration',ylabel='scores')
```

```python
df.plot(kind = 'bar',stacked=True,title='Ice cream Ratings',xlabel='Time Duration',ylabel='scores')
```

```python
df['Flavor Rating'].plot(kind = 'bar',title='Ice cream Ratings',xlabel='Time Duration',ylabel='scores')
```

```python
df.plot.barh(title='Ice cream Ratings',xlabel='scores',ylabel='Time Duration')
```

```python
df.plot.scatter(x='Texture Rating',y='Overall Rating',s=200,c='red')
```

```python
df.plot.hist(bins= 10)
```

```python
df.plot.box()
```

```python
df.plot.area(figsize=(10,5))
```

```python
df.plot.pie(y='Flavor Rating',figsize=(20,10))
```
