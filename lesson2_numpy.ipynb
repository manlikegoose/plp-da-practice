```python
before = np.array([[4,5,6],[7,2,5]])
print(before)

print('='*15)
after = before.reshape([,2])

print(after)
```

```python
#vertical stacking
v1 = np.array([3,4,5,7,2])
v2 = np.array([6,2,1,9,5])

# print(v1)
# print('='*15)
# print(v2)
v3 = np.vstack((v1,v2,v1))
print(v3)
```

```python
#horizontal
h1 = np.array([3,4,5,7,2])
h2 = np.array([6,2,1,9,5])

h3 = np.hstack((v1,v2))
print(h3)
```
