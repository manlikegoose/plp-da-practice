### Loading a dataset from datasets by importing load_dataset then using .to_padas to transform it into a dataframe

```python
dataset = load_dataset('lukebarousse/data_jobs')
```

## Data cleaning with pandas
#### - check the data types
#### - Handle Nan values
#### - sorting
#### - dropinf columns and nan

```python

df.info()
```

```python
#date format and conversion
df['job_posted_date'] = pd.to_datetime(df.job_posted_date)
```

```python
df.info()
```

```python
#creating a column based on the existing on
df['month_of_job_posted'] = df.job_posted_date.dt.month
```

```python
#sorting
df.sort_values(by = 'job_posted_date',inplace = True)
```

```python
df
```

```python
#droping columns using .drop

df.drop(labels='salary_hour_avg',axis=1,inplace=True)
```

```python
#Dropping Nan value using .dropna()

df.dropna(subset = ['salary_year_avg'])
```
