```python
import pandas as pd
```

```python
people = [
    {'fellowshipID':'1001','first_name':'John','last_name':'Raw','skills':'fishing'},
    {'fellowshipID':'1002','first_name':'Stark','last_name':'Tony','skills':'AI engineering'},
    {'fellowshipID':'1003','first_name':'Stan','last_name':'Smith','skills':'Robotics'},
    {'fellowshipID':'1004','first_name':'Hilla','last_name':'kiprotich','skills':'webdev'}
]


people
```

```python
df1= pd.DataFrame(people)
```

```python
people1 = [
    {'fellowshipID':'1001','first_name':'John','last_name':'Raw','Age':'32'},
    {'fellowshipID':'1002','first_name':'Stark','last_name':'Tony','Age':'45'},
    {'fellowshipID':'1006','first_name':'Sam','last_name':'soma','Age':'56'},
    {'fellowshipID':'1007','first_name':'kamu','last_name':'Seth','Age':'23'},
    {'fellowshipID':'1008','first_name':'Peter','last_name':'parker','Age':'61'}
]
```

```python
df2 = pd.DataFrame(people1)
df2
```

```python
#inner merge
df1.merge(df2,how='inner',on=['fellowshipID','first_name','last_name'])
```

```python
#outer merge
df1.merge(df2,how='outer',on=['fellowshipID','first_name','last_name'])
```

```python
#left merge
df1.merge(df2,how='left',on=['fellowshipID','first_name','last_name'])
```

```python
#right merge
df1.merge(df2,how='right',on=['fellowshipID','first_name','last_name'])
```

```python
df1.merge(df2,how='cross')
```

```python
df4 = df1.set_index('fellowshipID').join(df2.set_index('fellowshipID'),lsuffix='_left',rsuffix='_right',how='inner')
```
